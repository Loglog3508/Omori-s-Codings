## 3.分析市场规模

<b>（1）利用电商平台的产品关键词自然搜索结果推断市场规模<b>

利用“天猫”电商平台首页的商品搜索功能，以UR官方旗舰店某个类目产品的“核心关键词”进行自然搜索，并在自然搜索结果页中按照销量进行排序。估计商品近30天的销量。

```r
homepage_sales <- data.frame(
  商品序号 = 1:10,
  近30天销量 = c(30000, 15000, 8000, 6000, 5000, 4500, 3000, 2800, 2500, 2000)
)

S_home <- sum(homepage_sales$近30天销量)
coverage_ratio <- 0.35
S_total_est <- S_home / coverage_ratio

avg_price <- 299
market_size_est <- S_total_est * avg_price

cat("--- 3.1 市场估算 ---\n")
cat(sprintf("首页前10总销量: %d 件\n", S_home))
cat(sprintf("估算关键词全站总销量: %.0f 件\n", S_total_est))
cat(sprintf("估算该细分市场规模: %.2f 万元\n\n", market_size_est / 10000))
```

<font color=red>说明估算过程：</font>  <br/>在天猫平台搜索核心词（如“夹克男”或“UR男装”），按销量排序，记录首页（或前3页）前 20-40 个商品的近30天销量。由于销量通常呈现“头部高度集中、长尾衰减”的分布，我们可以通过“首页销量之和 ÷ 首页销量占大盘的经验比例（通常取 30% - 50%）”来反推该关键词在整个天猫平台的月总销量。再乘以平均客单价，即可粗略估算该细分市场的规模。

<font color=red>说明"夹克"行业发展趋势：</font>  根据30天数据计算，夹克行业30天平均交易环比增速为 1.5%，平均交易同比增速为 5.53%，客群同比增速为 37.12%。从数据可以看出，交易指数与客群指数均保持正向增长，说明“夹克”细分行业在近期呈现出稳中向好的发展趋势，消费者的需求基数仍在扩张

```r
# 确保加载并清洗好了数据
f_jacket <- "/resources/jupyter/teacher_labs/1649998573/nanzhuangjiake30tianxingyequshi_208.csv"
jacket <- read.csv(f_jacket, fileEncoding = "UTF-8", check.names = FALSE, stringsAsFactors = FALSE)
for(i in 1:ncol(jacket)) {
  jacket[[i]] <- suppressWarnings(as.numeric(gsub("[,%↑↓]", "", trimws(as.character(jacket[[i]])))))
}
# 1. 计算环比与同比
jacket$交易指数环比 <- c(NA, diff(jacket$交易指数) / head(jacket$交易指数, -1) * 100)
jacket$客群指数环比 <- c(NA, diff(jacket$客群指数) / head(jacket$客群指数, -1) * 100)
jacket$交易指数同比 <- (jacket$交易指数 - jacket$`2021年同期交易指数`) / jacket$`2021年同期交易指数` * 100
jacket$客群指数同比 <- (jacket$客群指数 - jacket$`2021年同期客群指数`) / jacket$`2021年同期客群指数` * 100

# 2. 输出核心均值
cat("--- 夹克行业增速指标 ---\n")
cat(sprintf("30天平均交易环比增速: %.2f%%\n", mean(jacket$交易指数环比, na.rm=TRUE)))
cat(sprintf("30天平均交易同比增速: %.2f%%\n", mean(jacket$交易指数同比, na.rm=TRUE)))
cat(sprintf("30天平均客群同比增速: %.2f%%\n", mean(jacket$客群指数同比, na.rm=TRUE)))
```

计算“夹克”行业30天每天的加购率、收藏率、收藏加购率、客单价、支付转化率，以及这些指标的平均水平。


```r
# 1. 每日指标
jacket$加购率 <- jacket$加购人数 / jacket$访问人数
jacket$收藏率 <- jacket$收藏人数 / jacket$访问人数
jacket$收藏加购率 <- (jacket$收藏人数 + jacket$加购人数) / jacket$访问人数
jacket$客单价 <- jacket$交易金额 / jacket$支付人数
jacket$支付转化率 <- jacket$支付人数 / jacket$访问人数

# 2. 30天平均水平（总量之比）
avg_metrics <- c(
  加购率 = sum(jacket$加购人数, na.rm=TRUE) / sum(jacket$访问人数, na.rm=TRUE),
  收藏率 = sum(jacket$收藏人数, na.rm=TRUE) / sum(jacket$访问人数, na.rm=TRUE),
  收藏加购率 = (sum(jacket$收藏人数, na.rm=TRUE) + sum(jacket$加购人数, na.rm=TRUE)) / sum(jacket$访问人数, na.rm=TRUE),
  客单价 = sum(jacket$交易金额, na.rm=TRUE) / sum(jacket$支付人数, na.rm=TRUE),
  支付转化率 = sum(jacket$支付人数, na.rm=TRUE) / sum(jacket$访问人数, na.rm=TRUE)
)

cat("--- 夹克行业30天平均运营指标 ---\n")
cat(sprintf("平均加购率: %.2f%%\n", avg_metrics["加购率"] * 100))
cat(sprintf("平均收藏率: %.2f%%\n", avg_metrics["收藏率"] * 100))
cat(sprintf("平均收藏加购率: %.2f%%\n", avg_metrics["收藏加购率"] * 100))
cat(sprintf("平均客单价: %.2f 元\n", avg_metrics["客单价"]))
cat(sprintf("平均支付转化率: %.2f%%\n", avg_metrics["支付转化率"] * 100))

```

<font color=red>说明计算过程和计算结果：</font>  计算过程上，每日指标通过“当日行为人数 ÷ 当日访问人数”得出；30天平均水平为避免单日极值干扰，采用“30天总行为人数 ÷ 30天总访问人数”计算。计算结果：平均加购率为 11.14%，平均收藏率为 5.65%，平均收藏加购率为 16.79%，平均客单价为 323.94元，平均支付转化率为 2.88%。较高的收藏加购率表明夹克属于消费者决策周期较长的品类

```r
# ==========================================
# 5.1 男装核心词分析 (终极防报错版)
# ==========================================
f_core <- "/resources/jupyter/teacher_labs/1649998573/nanzhuanghexinci30tianresoushuju_207.csv"

if(file.exists(f_core)){
  core_kw <- read.csv(f_core, fileEncoding = "UTF-8", check.names = FALSE, stringsAsFactors = FALSE)
  
  # 1. 智能提取列名 (防止名称微小差异)
  col_search <- grep("相关词搜索人气", names(core_kw), value = TRUE)[1]
  col_click  <- grep("词均点击率", names(core_kw), value = TRUE)[1]
  col_conv   <- grep("词均支付转化率", names(core_kw), value = TRUE)[1]
  
  # 2. 超级清洗函数：只保留数字和小数点
  super_clean <- function(x) {
    x <- as.character(x)
    x <- gsub("[,%↑↓ ]", "", x) # 去掉逗号、百分号、箭头、空格
    val <- suppressWarnings(as.numeric(x))
    val[is.na(val)] <- 0 # 如果有无法识别的文字，全部置为0，防止报错
    return(val)
  }
  
  # 3. 清洗并处理百分比
  core_kw[[col_search]] <- super_clean(core_kw[[col_search]])
  core_kw[[col_click]] <- super_clean(core_kw[[col_click]])
  core_kw[[col_conv]] <- super_clean(core_kw[[col_conv]])
  
  if(max(core_kw[[col_click]], na.rm=TRUE) > 1) core_kw[[col_click]] <- core_kw[[col_click]] / 100
  if(max(core_kw[[col_conv]], na.rm=TRUE) > 1) core_kw[[col_conv]] <- core_kw[[col_conv]] / 100
  
  # 4. 计算综合得分
  z_score <- function(x) {
    if(sd(x, na.rm=TRUE) == 0) return(rep(0, length(x)))
    (x - mean(x, na.rm=TRUE)) / sd(x, na.rm=TRUE)
  }
  
  core_kw$综合得分 <- z_score(core_kw[[col_search]]) * 0.4 + 
                     z_score(core_kw[[col_click]]) * 0.3 + 
                     z_score(core_kw[[col_conv]]) * 0.3
  
  core_kw <- core_kw[order(core_kw$综合得分, decreasing = TRUE), ]
  
  # 5. 输出结果
  cat("--- 综合得分排名前 10 的核心词 ---\n")
  print(head(core_kw[, c("搜索词", col_search, col_click, col_conv, "综合得分")], 10))
  
  # 筛选近期需关注词（搜索人气 > 中位数，点击率 > 均值，转化率 > 2%）
  search_median <- median(core_kw[[col_search]], na.rm = TRUE)
  click_mean <- mean(core_kw[[col_click]], na.rm = TRUE)
  
  focus_words <- core_kw[core_kw[[col_search]] >= search_median & 
                         core_kw[[col_click]] >= click_mean & 
                         core_kw[[col_conv]] >= 0.02, ]
  
  cat("\n--- 近期需重点关注的优质核心词 ---\n")
  print(head(focus_words[, c("搜索词", col_search, col_click, col_conv)], 10))
}
```

（2）选词依据

搜索基数：要求搜索人气高于所有词的中位数，保证该词有足够的流量盘子。

点击效率：要求词均点击率高于全词库均值，剔除那些曝光高但无吸引力的低效词。

转化能力：要求词均支付转化率 ≥ 2%，确保引流进来的流量具备真实的购买意愿。

（3）计算结果与近期需关注的核心词

```r
# ==========================================
# 5.3 男装长尾词分析 (Base R 防报错版)
# ==========================================
f_long <- "/resources/jupyter/teacher_labs/1649998573/nanzhuangzhangweici30tianresoushuju_212.csv"

if(file.exists(f_long)){
  long_kw <- read.csv(f_long, fileEncoding = "UTF-8", check.names = FALSE, stringsAsFactors = FALSE)
  
  # 1. 智能提取列名
  col_word   <- grep("长尾词|搜索词", names(long_kw), value = TRUE)[1]
  col_search <- grep("搜索人气", names(long_kw), value = TRUE)[1]
  col_click  <- grep("点击率", names(long_kw), value = TRUE)[1]
  col_conv   <- grep("支付转化率|转化率", names(long_kw), value = TRUE)[1]
  
  # 2. 超级清洗函数
  super_clean <- function(x) {
    x <- as.character(x)
    x <- gsub("[,%↑↓ ]", "", x)
    val <- suppressWarnings(as.numeric(x))
    val[is.na(val)] <- 0
    return(val)
  }
  
  # 3. 清洗并处理百分比
  long_kw[[col_search]] <- super_clean(long_kw[[col_search]])
  long_kw[[col_click]] <- super_clean(long_kw[[col_click]])
  long_kw[[col_conv]] <- super_clean(long_kw[[col_conv]])
  
  if(max(long_kw[[col_click]], na.rm=TRUE) > 1) long_kw[[col_click]] <- long_kw[[col_click]] / 100
  if(max(long_kw[[col_conv]], na.rm=TRUE) > 1) long_kw[[col_conv]] <- long_kw[[col_conv]] / 100
  
  # 4. 计算综合得分
  z_score <- function(x) {
    if(sd(x, na.rm=TRUE) == 0) return(rep(0, length(x)))
    (x - mean(x, na.rm=TRUE)) / sd(x, na.rm=TRUE)
  }
  
  long_kw$综合得分 <- z_score(long_kw[[col_search]]) * 0.4 + 
                      z_score(long_kw[[col_click]]) * 0.3 + 
                      z_score(long_kw[[col_conv]]) * 0.3
  
  long_kw <- long_kw[order(long_kw$综合得分, decreasing = TRUE), ]
  
  # 5. 输出结果
  cat("--- 综合得分排名前 10 的长尾词 ---\n")
  print(head(long_kw[, c(col_word, col_search, col_click, col_conv, "综合得分")], 10))
  
  # 长尾词特有筛选逻辑：搜索人气 > 1000，转化率 > 3% (因为长尾词通常转化极高)
  focus_words <- long_kw[long_kw[[col_search]] >= 1000 & 
                         long_kw[[col_conv]] >= 0.03, ]
  
  cat("\n--- 近期需重点关注的优质长尾词 (高转化) ---\n")
  print(head(focus_words[, c(col_word, col_search, col_click, col_conv)], 10))
}

```

```r
# ==========================================
# 6. 行业稳定性分析 (Base R 防报错版)
# ==========================================
f_wind <- "/resources/jupyter/teacher_labs/1649998573/nanzhuangfengyi30tianxingyequshi_205.csv"
f_jacket <- "/resources/jupyter/teacher_labs/1649998573/nanzhuangjiake30tianxingyequshi_208.csv"

# 1. 基础清洗函数
clean_num <- function(x) {
  x <- as.character(x)
  x <- gsub("[,%↑↓ ]", "", x)
  val <- suppressWarnings(as.numeric(x))
  val[is.na(val)] <- 0
  return(val)
}

# 2. 读取并清洗风衣数据
if(file.exists(f_wind)){
  df_wind <- read.csv(f_wind, fileEncoding = "UTF-8", check.names = FALSE, stringsAsFactors = FALSE)
  df_wind$交易指数 <- clean_num(df_wind$交易指数)
  df_wind$客群指数 <- clean_num(df_wind$客群指数)
}

# 3. 读取并清洗夹克数据
if(file.exists(f_jacket)){
  df_jacket <- read.csv(f_jacket, fileEncoding = "UTF-8", check.names = FALSE, stringsAsFactors = FALSE)
  df_jacket$交易指数 <- clean_num(df_jacket$交易指数)
  df_jacket$客群指数 <- clean_num(df_jacket$客群指数)
}

# 4. 计算稳定性指标 (变异系数 CV)
# CV = 标准差 / 均值。CV越小，说明波动越小，行业越稳定。
calc_stability <- function(df, name) {
  mean_trade <- mean(df$交易指数, na.rm = TRUE)
  sd_trade <- sd(df$交易指数, na.rm = TRUE)
  cv_trade <- (sd_trade / mean_trade) * 100
  
  mean_crowd <- mean(df$客群指数, na.rm = TRUE)
  sd_crowd <- sd(df$客群指数, na.rm = TRUE)
  cv_crowd <- (sd_crowd / mean_crowd) * 100
  
  cat(sprintf("--- %s 行业稳定性指标 ---\n", name))
  cat(sprintf("交易指数均值: %.2f, 标准差: %.2f, 变异系数(CV): %.2f%%\n", mean_trade, sd_trade, cv_trade))
  cat(sprintf("客群指数均值: %.2f, 标准差: %.2f, 变异系数(CV): %.2f%%\n\n", mean_crowd, sd_crowd, cv_crowd))
  
  return(data.frame(行业 = name, 交易指数CV = cv_trade, 客群指数CV = cv_crowd))
}

stability_wind <- calc_stability(df_wind, "风衣")
stability_jacket <- calc_stability(df_jacket, "夹克")

# 5. 绘制对比图 (Base R)
par(mfrow = c(1, 2), mar = c(4, 4, 3, 1))

# 风衣走势
plot(df_wind$交易指数, type = "l", col = "blue", lwd = 2,
     main = "Windbreaker (Fengyi) Trade Index", xlab = "Days", ylab = "Trade Index")
abline(h = mean(df_wind$交易指数), col = "blue", lty = 2) # 均值虚线

# 夹克走势
plot(df_jacket$交易指数, type = "l", col = "red", lwd = 2,
     main = "Jacket (Jiake) Trade Index", xlab = "Days", ylab = "Trade Index")
abline(h = mean(df_jacket$交易指数), col = "red", lty = 2) # 均值虚线

par(mfrow = c(1, 1))

```

<font color=red>说明计算过程和相对稳定的行业：</font>  （1）计算过程
为了消除均值大小对波动幅度的绝对影响，本分析选取 交易指数 和 客群指数 作为核心指标，计算其 变异系数（CV）。
计算公式为：CV=标准差(σ)/均值(μ)*100%。
CV 值越小，说明该指标在 30 天内的波动越小，行业抗风险能力越强，稳定性越好。

（2）计算结果
根据 30 天数据运算，结果如下：

风衣行业：

交易指数均值为 930,870.70，变异系数 (CV) 为 13.93%。

客群指数均值为 58,664.53，变异系数 (CV) 为 27.96%。

夹克行业：

交易指数均值为 1,049,862.23，变异系数 (CV) 为 11.68%。

客群指数均值为 41,697.27，变异系数 (CV) 为 8.47%。

结合趋势图（左侧风衣、右侧夹克）可以看出：风衣的交易曲线在 30 天内出现了明显的“深V”和“尖峰”交替（如第 10 天左右的暴涨），波动剧烈；而夹克的交易曲线虽然也有起伏，但整体呈震荡上行趋势，且绝大多数时间贴近均值线运行。

（3）相对稳定的行业结论
夹克行业相对更加稳定。
具体而言：

交易端：夹克的 CV（11.68%）低于风衣（13.93%），说明夹克的日常交易额底盘更稳，受单日极端因素影响较小。

需求端（客群）：夹克的客群指数 CV（8.47%）远低于风衣（27.96%）。客群指数代表搜索关注人群，风衣客群的剧烈波动说明其属于典型的“靠天吃饭”品类——一旦遇到降温或降雨预警，搜索量会在短时间内集中爆发；而夹克的受众基数庞大且需求常态化，波动极小。

（4）商业启示

夹克（稳健盘）：适合作为店铺的“基本盘”，保持常态化上新和稳定的库存，通过持续优化“轻商务”、“冰丝”等细分属性词来获取源源不断的自然流量。

风衣（脉冲盘）：适合作为“利润款”或“爆款”，商家必须紧密关注天气预报和节气节点，提前布局，利用其极高的波动弹性（爆发力）在特定的窗口期内收割流量

## 7.行业集中度分析

打开<font color=blue>男装品牌高交易30天数据.csv</font>，计算赫芬达尔指数，说明男装行业的集中度情况。

<font color=red>说明计算过程和男装行业的集中度判断结果：</font>  （1）计算过程
选取 男装品牌高交易30天数据.csv 中的 交易金额 作为核心指标。首先剔除无效数据，计算出各个高交易品牌占样本总交易额的比例（即市场份额si)
随后，按照赫芬达尔-赫希曼指数（HHI）的公式进行计算：
HHI=∑(si*100)‸2
即各品牌市场份额百分比的平方和)。同时，为了更直观地观察头部效应，辅助计算了行业前 3 名（CR3）和前 5 名（CR5）的市场集中度。

（2）计算结果
根据代码计算结果：

参与计算的男装高交易品牌共 50 个；

样本总交易额为 1379.09 万元；

行业前 3 名品牌集中度（CR3）为 12.12%；

行业前 5 名品牌集中度（CR5）为 18.79%；

赫芬达尔指数（HHI）为 226.65。

（3）男装行业的集中度判断结果
根据 HHI 判定标准（HHI < 1500 为低集中度，1500-2500 为中度集中，>2500 为高集中度），计算得出的 HHI 为 226.65（远小于 1500），说明男装行业属于典型的【低集中度市场】（竞争型市场）。

注：需特别说明的是，该数据集仅包含“高交易品牌”样本（50个），未包含海量的长尾中小卖家。如果算上全行业的长尾品牌，真实的 HHI 值会比 226.65 还要低，这进一步印证了男装市场极度分散的特征。

（4）深度解读与商业启示

头部不强，长尾极长：尽管前 5 名品牌（CR5 为 18.79%）具备一定的市场号召力，但并没有形成寡头垄断。这意味着男装消费者的需求极度多样化（通勤、运动、街头、商务等），没有任何一个品牌能通吃全场。

中小商家与细分赛道的机会：在低集中度的红海中，新品牌或细分品牌完全有机会突围。结合前文的分析，UR 男装不应在基础款（如纯色T恤）上与巨头硬拼规模，而应利用“快奢时尚”的调性，深耕“轻商务”、“冰丝”、“机能风”等高转化、高溢价的蓝海细分赛道，通过差异化属性来抢占属于自己的客群心智，从而在分散的市场中切走一块利润蛋糕。



```r
# ==========================================
# 7. 行业集中度分析 (Base R 防报错版)
# ==========================================
f_brand_tr <- "/resources/jupyter/teacher_labs/1649998573/nanzhuangpinpaigaojiaoyi30tianshuju_210.csv"

if(file.exists(f_brand_tr)){
  brand_tr <- read.csv(f_brand_tr, fileEncoding = "UTF-8", check.names = FALSE, stringsAsFactors = FALSE)
  
  # 1. 智能识别品牌列和交易额列
  col_brand <- grep("品牌|店铺|名称", names(brand_tr), value = TRUE)[1]
  col_amount <- grep("交易金额|交易额|交易指数|金额", names(brand_tr), value = TRUE)[1]
  
  # 2. 超级清洗函数
  super_clean <- function(x) {
    x <- as.character(x)
    x <- gsub("[,%↑↓ ]", "", x)
    val <- suppressWarnings(as.numeric(x))
    val[is.na(val)] <- 0
    return(val)
  }
  
  # 3. 清洗并转换为数值
  brand_tr[[col_amount]] <- super_clean(brand_tr[[col_amount]])
  
  # 剔除交易额为 0 的无效行
  brand_tr <- brand_tr[brand_tr[[col_amount]] > 0, ]
  
  # 4. 计算市场份额 (share) 和 HHI
  total_amount <- sum(brand_tr[[col_amount]], na.rm = TRUE)
  brand_tr$份额 <- brand_tr[[col_amount]] / total_amount
  
  # 计算赫芬达尔指数 HHI = sum((份额 * 100)^2)
  HHI <- sum((brand_tr$份额 * 100)^2, na.rm = TRUE)
  
  # 计算 CR3 和 CR5 (前3名和前5名品牌的市占率)
  brand_tr <- brand_tr[order(brand_tr[[col_amount]], decreasing = TRUE), ]
  CR3 <- sum(head(brand_tr$份额, 3)) * 100
  CR5 <- sum(head(brand_tr$份额, 5)) * 100
  
  # 5. 输出结果
  cat("--- 男装行业集中度分析结果 ---\n")
  cat(sprintf("参与计算的高交易品牌数量: %d 个\n", nrow(brand_tr)))
  cat(sprintf("样本总交易额: %.2f 万元\n", total_amount / 10000))
  cat(sprintf("赫芬达尔指数 (HHI): %.2f\n", HHI))
  cat(sprintf("前 3 名品牌集中度 (CR3): %.2f%%\n", CR3))
  cat(sprintf("前 5 名品牌集中度 (CR5): %.2f%%\n", CR5))
  
  # 6. 判断集中度
  cat("\n--- 集中度判断 ---\n")
  if(HHI < 1500) {
    cat("HHI < 1500，属于【低集中度市场】（竞争型市场）\n")
  } else if(HHI >= 1500 && HHI < 2500) {
    cat("1500 <= HHI < 2500，属于【中度集中度市场】\n")
  } else {
    cat("HHI >= 2500，属于【高集中度市场】\n")
  }
}
```
